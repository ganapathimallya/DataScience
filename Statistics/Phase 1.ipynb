```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import statistics as st
```

```python
titanicData = pd.read_csv("Dataset/titanic.csv")
display(titanicData.head(5))
```

## Mean (Avg)

```python
meanValue = titanicData["Fare"].mean()
meanValue
```

```python

sns.histplot(x="Fare",data=titanicData,bins=[i for i in range(0,150,10)])
plt.plot([meanValue for i in range(0,300)],[i for i in range(0,300)],c="Red",label="Mean")
plt.show()
```

## Median

```python
medianValue = np.median(titanicData["Fare"])
medianValue
```

```python
sns.histplot(x="Fare",data=titanicData,bins=[i for i in range(0,150,10)])
plt.plot([medianValue for i in range(0,300)],[i for i in range(0,300)],c="Green",label="Mean")
plt.show()
```

## Mode

```python
modeValue = st.mode(titanicData["Fare"])
modeValue
```

```python
sns.histplot(x="Fare",data=titanicData,bins=[i for i in range(0,150,10)])
plt.plot([modeValue for i in range(0,370)],[i for i in range(0,370)],c="Orange",label="mode")
plt.ylim(0,400)
plt.show()
```

## Fianl Output

```python
#Final plot
sns.histplot(x="Fare",data=titanicData,bins=[i for i in range(0,150,10)])
plt.plot([meanValue for i in range(0,300)],[i for i in range(0,300)],c="Red",label="Mean")
plt.plot([medianValue for i in range(0,300)],[i for i in range(0,300)],c="Green",label="Median")
plt.plot([modeValue for i in range(0,370)],[i for i in range(0,370)],c="Orange",label="mode")
plt.ylim(0,400)
plt.legend()
plt.show()
```
