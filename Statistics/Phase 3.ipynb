```python
student_df=pd.read_csv("Dataset/student_scores.csv")
student_df['ReadingScore'].mean(),student_df['ReadingScore'].median(),st.mode(student_df['ReadingScore'])
student_df.head(5)
```

```python
sns.histplot(data=student_df,x='MathScore',bins=10)
plt.axvline(student_df['MathScore'].mean(),color='red')
plt.axvline(student_df['MathScore'].median(),color='blue')
plt.axvline(st.mode(student_df['MathScore']),color='green')
plt.show()
```

```python
groupA_df=student_df[student_df['EthnicGroup']=='group A']
max_m=groupA_df['MathScore'].max()
min_m=groupA_df['MathScore'].min()
range_mathscore=max_m-min_m
max_m,min_m,range_mathscore
```

```python
male_df=student_df[student_df['Gender']=='male']
female_df=student_df[student_df['Gender']=='female']
mean_male=male_df['MathScore'].mean()
mad_male=np.sum(np.abs(male_df['MathScore']-mean_male))/len(male_df['MathScore'])
mean_female=female_df['MathScore'].mean()
mad_female=np.sum(np.abs(female_df['MathScore']-mean_female))/len(female_df['MathScore'])
mad_male,mad_female
```
