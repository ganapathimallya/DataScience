# Coverience and Corelation

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
sales_df=pd.read_csv("Dataset/monthly_sales.csv")
sales_df.head()
#Filtering the data for January sales
sales_jan=sales_df[sales_df['Sales Month']=='Jan']
sales_jan.head(2)
#Filter the data for Feb sales
sales_feb=sales_df[sales_df['Sales Month']=='Feb']
sales_feb.head(2)
#Filter the data for March sales
sales_mar=sales_df[sales_df['Sales Month']=='Mar']
sales_mar.head(2)
```

```python
# No correlation between Advertising and Sales in January sales data
sales_jan['Advertising'].corr(sales_jan['Sales'])
sns.scatterplot(data=sales_jan,x='Advertising',y='Sales')
plt.title("January Sales Correlation")
plt.show()
```

```python
# Positive correlation between Advertising and Sales in February sales data

sales_feb['Advertising'].corr(sales_feb['Sales'])
sns.scatterplot(data=sales_feb,x='Advertising',y='Sales')
plt.title("February Sales Correlation")
plt.show()
```

```python
# Negative correlation between Advertising and Sales in March sales data
sales_mar['Advertising'].corr(sales_mar['Sales'])
sns.scatterplot(data=sales_mar,x='Advertising',y='Sales')
plt.title("March Sales Correlation")
plt.show()
```
