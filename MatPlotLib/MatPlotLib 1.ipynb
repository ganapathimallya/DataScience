## MathPlotLib

```python
import matplotlib.pyplot as plt
import pandas as pd
```

### Bar Plot

```python
cname=["Java","Python","C","C++"]
enrollment=[24,30,12,14]
plt.bar(cname,enrollment)
```

```python
course_data={
    "Caname":["Java","Python","C","C++"],
    "Enrollment":[24,30,12,14]
}
cource_df=pd.DataFrame(course_data)
bars=plt.bar(
    cource_df["Caname"],
    cource_df["Enrollment"],
    color=["#00FFEAFF","#4AD34A","#F73126","#1F22FF"],
    edgecolor="black",
    width=0.8
)
#Add title and labels
plt.title("Course Enrollment Overview",fontsize=20,pad=15,fontweight="bold")
plt.xlabel("Course",fontsize=16)
plt.ylabel("Enrollment",fontsize=16)
plt.bar_label(bars,fontsize=12,padding=0)
plt.ylim(0,40)
plt.gca().spines["top"].set_visible(False)
plt.gca().spines["right"].set_visible(False)
plt.grid(axis="y",linestyle="--",alpha=0.7)
plt.tight_layout()
plt.show()
```

```python
data=pd.read_excel("expense.xlsx")
exp_df=pd.DataFrame(data)
expgrouped=exp_df.groupby("Category")["Amount"].sum()
plt.bar(
    expgrouped.index,
    expgrouped.values,
    color=["#073F88FF","#E5FF00","#D40B00","#C600B6"],
    edgecolor="black",
    width=0.65
)
plt.title("Expense Overview",fontsize=20,pad=15,fontweight="bold")
plt.xlabel("Category",fontsize=16)
plt.ylabel("Amount",fontsize=16)
```

## Line Plot

```python
weekly_Sales={
    "Day":["Mon","Tue","Wed","Thu","Fri","Sat","Sun"],
    "Week1":[15,12,18,8,10,19,30],
    "Week2":[25,14,10,18,13,17,35],
}
sales=pd.DataFrame(weekly_Sales)
plt.plot(
    sales["Day"],
    sales["Week1"],
    marker="o",
    ls="--",
    label="Week1_Sales",
    color="#1900FF",
    markersize=10,
    linewidth=2
)
plt.plot(
    sales["Day"],
    sales["Week2"],
    marker="o",
    ls="-",
    label="Week2_Sales",
    color="#FF0000",
    markersize=10,
    linewidth=2
)
plt.title("Weekly Sales Overview",fontsize=20,pad=15,fontweight="bold")
plt.xlabel("Day",fontsize=16)
plt.ylabel("Sales",fontsize=16)
plt.show()
```

## Scatter Plot

```python
adv_sales={
    "Advertising":[10,20,25,25,25,20,29,35],
    "Sales":[130,280,300,230,340,400,250,360]
}
sales_df=pd.DataFrame(adv_sales)
plt.scatter(
    sales_df["Advertising"],
    sales_df["Sales"],
    color="red",
    marker="*",
    alpha=0.7,
    s=200,
    label="Sales vs Advertising"
)
plt.title("Sales vs Advertising",fontsize=20,pad=15,fontweight="bold")
plt.xlabel("Advertising",fontsize=16)
plt.ylabel("Sales",fontsize=16)
plt.show()
```
