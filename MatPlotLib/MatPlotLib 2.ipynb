```python
import pandas as pd
import matplotlib.pyplot as plt
```

## Pie Chart

```python
monthly_data={
    "brand":["OnePlus","Apple","Vivo","Redmi","Oppo"],
    "sales":[26,7,19,23,27]
}
sales_df=pd.DataFrame(monthly_data)
sales_df
plt.figure(figsize=(7,6))
plt.pie(
    sales_df["sales"],
    labels=sales_df["brand"],
    colors=["#f99e00","#ff0000","#00b3ff","#e8790b","#006b25"],
    explode=[0,0,0,0,0.1],
    shadow=True,
    startangle=180,
    autopct="%.2f%%",
    normalize=True
    )
plt.title("Mobile Sales Distribution",fontsize=12*2,fontweight="bold")
plt.legend(sales_df["brand"],title="Mobile Brands")
plt.tight_layout()
plt.show()
```

```python
## Pie chart for expense with upi,cash,and card
data=pd.read_excel("expense.xlsx")
exp_df=pd.DataFrame(data)
expgrouped=exp_df.groupby("Payment Mode")["Amount"].sum()
print(expgrouped)
plt.figure(figsize=(6,8))
plt.pie(
    expgrouped.values,
    colors=["#073F88FF","#E5FF00","#D40B00","#C600B6"],
    labels=expgrouped.index,
    shadow=True,
    explode=[0,0.05,0],
    autopct="%.2f%%"
)
plt.title("Expense Overview",fontsize=20,pad=15,fontweight="bold")
plt.legend(expgrouped.index)
```

## Histogram

```python
wether={
    "temp":[33,30,30,32,33,34,32,32,32,33,33,32,32,34],
    "time":['9.00','9.30','10.00','10.30','11.00','11.30','12.00','12.30','13.00','13.30','14.00','14.30','15.00','15.30']
}
wether_df=pd.DataFrame(wether)
plt.hist(
    wether_df["temp"],
    bins=6,
    color="skyblue",
    edgecolor="Black",
    alpha=0.5
)
plt.show()
```

## Violin Plot

```python
import matplotlib.pyplot as plt

class_a = [10,20,30,40,41,42,43,50,60,70,71,73,80,90,100]
plt.violinplot(
    [class_a],
    showextrema=True,
    showmeans=True,
    showmedians=True
    )
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

weekly_sales = {
    "Days": ["Sun", "Mon", "Tue", "Wed", "Thu", "Fri", "Sat"],
    "Week1": [5, 10, 15, 20, 25, 30, 40],
    "Week2": [10, 20, 30, 40, 45, 50, 55],
    "Week3": [10, 40, 45, 46, 48, 50, 55]
}

sales_df = pd.DataFrame(weekly_sales)

plt.stackplot(
    sales_df["Days"],
    sales_df["Week1"],
    sales_df["Week2"],
    sales_df["Week3"],
    labels=["Week1", "Week2", "Week3"]
)
plt.legend()
plt.show()
```
