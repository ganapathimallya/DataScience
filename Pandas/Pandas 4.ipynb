## Group By and Join

```python
import pandas as pd
import numpy as np
```

```python
exp=pd.read_excel('Subdata_pandas/expense.xlsx')
exp
```

```python
#Total expences for each catogary
#M1
exp.groupby('Category')["Amount"].sum()
#M2
exp.groupby('Category')["Amount"].agg(['sum'])
```

```python
exp.groupby('Category')["Amount"].agg(['sum','count','mean','max','min'])
```

```python
#how much money paid by upi cash and card
exp.groupby('Payment Mode')['Amount'].agg(['sum','count'])
```

```python
#For each cotegory how much money paid by upi cash and card
exp.groupby(['Payment Mode','Category'])['Amount'].agg(['sum','count'])
```

```python
#Which category has the highest expences
exp.groupby('Category')['Amount'].agg('sum').sort_values(ascending=False)

exp.groupby('Category')['Amount'].sum().sort_values(ascending=False).reset_index
```

```python
#HW
# create a dataframe containing the total expences for each category

total_expence=exp.groupby('Category')["Amount"].agg(['sum'])
total_expence

# What is the avgerage transaction amount for each payment mode

exp.groupby('Payment Mode')['Amount'].agg('mean')


# Find the highest individual trasaction in each category

exp.groupby('Category')['Amount'].agg('max')

# show total,avg,min,max trasction amount for each payment mode

exp.groupby('Payment Mode')['Amount'].agg(['sum','mean','min','max'])


```
