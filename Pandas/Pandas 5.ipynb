## Compare Pivoting and Melting DataFrame

```python
import pandas as pd
```

### Compare

```python
shop_7Oct={
    "Fruits":['Mango','Apple','Banana','Papaya'],
    "Price":[150,200,60,25],
    "Quantity":[15,7,12,13]
}
stock_7Oct=pd.DataFrame(shop_7Oct)

shop_8Oct={
    "Fruits":['Mango','Apple','Banana','Papaya'],
    "Price":[200,250,60,35],
    "Quantity":[10,17,24,15]
}
stock_8Oct=pd.DataFrame(shop_8Oct)
```

```python
stock_7Oct.compare(stock_8Oct)
stock_7Oct.compare(stock_8Oct, align_axis=0)
stock_7Oct.compare(stock_8Oct,keep_equal=True)
stock_7Oct.compare(stock_8Oct,keep_shape=True)
stock_7Oct.compare(stock_8Oct,keep_shape=True,keep_equal=True)
```

### Pivoting

```python
monthly_shop={
    "Product":['Laptop','Mobile','Laptop','Mobile','Laptop','Laptop'],
    "Month":['Jan','Jan','Feb','Feb','Mar','Apr'],
    "Price":[50000,27500,65000,45000,95000,35000]
}
monthly_sales=pd.DataFrame(monthly_shop)
monthly_sales

monthly_sales.pivot(
    index="Product",
    columns="Month",
    values="Price"
)
```

```python
monthly_shop={
    "Product":['Laptop','Mobile','Laptop','Mobile','Laptop','Laptop'],
    "Month":['Jan','Jan','Feb','Feb','Mar','Mar'],
    "Price":[50000,27500,65000,45000,95000,35000]
}
monthly_sales=pd.DataFrame(monthly_shop)
monthly_sales
monthly_sales.pivot_table(
    index="Product",
    columns="Month",
    values="Price",
    aggfunc="sum"
)
```

### Melting

```python
shop_sales={
    "Product":['Laptop','Mobile','Mouse'],
    "Jan":[67500,25000,950],
    "Feb":[87000,35000,850],
    "Mar":[98000,23500,650]
}
sales=pd.DataFrame(shop_sales)
sales.melt(
    id_vars="Product",
    var_name="Month",
    value_name="Sales"
)
```
