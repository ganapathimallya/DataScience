## Dealing with Duplicate data and Missing value

### Duplicate Data

```python
cdata=pd.read_csv("Subdata_pandas/company.csv")
cdata
```

```python
cdata.duplicated()
#cheack emp id
cdata["EEID"].duplicated()
#cheak Name
cdata["Name"].duplicated()
#Drop data
cdata['EEID'].drop_duplicates()
```

### Missing Value

```python
cdata.isnull().sum()
#Eliminate null values
cdata.dropna() #Drop all row containg null value
```

#### METHOD 1

```python
#Replace null with any value
cdata.replace(np.nan,"EmptyData")
mean=np.mean(cdata["salary"])
#Replace null in salary column with mean of salary
cdata["salary"].replace(np.nan,mean)
```

#### METHOD 2

```python
cdata['salary']=cdata['salary'].fillna(mean)
cdata
```

```python
cdata=pd.read_csv("Subdata_pandas/company.csv")
print(cdata)
# Forword fill - previous data will be filled
# cdata.ffill()
#Backword fill - next data will be filled
# cdata.bfill()
cdata["salary"]=cdata['salary'].bfill()
cdata
```
