```python
fruits=["apple", "banana", "cherry"]
print(fruits[::-1])
fruits.append("orange")  #adding element at the end of the list
print(fruits)
fruits.pop()  #removing last element from the list
print(fruits)
#to remove a specific element from the list we can use remove() method
fruits.remove("banana")
print(fruits)
```

```python
nums=[4,3,5,2,1,6]
nums.sort()  #sorts the list in ascending order
print(nums)
```

```python
student=["Ganapathi", 20, 8.83,"Mangalore"]
for i in range(len(student)):
    print(student[i],end=" ")
print("\n")
for val in student:
    print(val,end="\t")
```
