## NumPy Array Operation

```python
arr=np.array([[1,2,3,4,5]])
print("Slicing:\n",arr[:5],"\n")
print("-Ve Slicing:\n",arr[::],"\n")
```

```python
#Sales year wise id,2023,2024,2025
sales=np.array([[10,1000,1200,900],#Noodles
                [20,1100,1300,1000],#Pizza
                [30,1200,1400,1100]])#Icecream
print("Pizza Sales 2024:",sales[1,2],"\n")

print("Noodles Sales 2025:",sales[0,1:],"\n")

print("Sales of all items in 2024:",sales[:,1],"\n")

print("Sales data for 2023,2024:",sales[1,1:3],"\n")

print("Last two years sales for pizza and icecream:\n",sales[1:,2:],"\n")
```

## Fancy Indexing and `np.where`

**Fancy indexing** selects multiple elements using a list or array of indices. `np.where` can return matching indices or choose values based on a condition.
import numpy as np

# Fancy indexing with a 1D array
scores = np.array([72, 91, 65, 88, 79])
selected = scores[[1, 3, 4]]
print("Scores at positions 1, 3, and 4:", selected)

# Fancy indexing with a 2D array: select rows and columns together
sales = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])
print("Selected rows 0 and 2:\n", sales[[0, 2]])
print("Selected values at (row, column) pairs:", sales[[0, 2], [1, 0]])

# np.where(condition) returns the indices where the condition is True
even_positions = np.where(scores % 2 == 0)
print("Positions of even scores:", even_positions[0])
print("Even scores:", scores[even_positions])

# np.where(condition, value_if_true, value_if_false) replaces values by condition
result = np.where(scores >= 80, "Pass", "Needs improvement")
print("Result:", result)

```python
import numpy as np

# Fancy indexing with a 1D array
scores = np.array([72, 91, 65, 88, 79])
selected = scores[[1, 3, 4]]
print("Scores at positions 1, 3, and 4:", selected)

# Fancy indexing with a 2D array: select rows and columns together
sales = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])
print("Selected rows 0 and 2:\n", sales[[0, 2]])
print("Selected values at (row, column) pairs:", sales[[0, 2], [1, 0]])

# np.where(condition) returns the indices where the condition is True
even_positions = np.where(scores % 2 == 0)
print("Positions of even scores:", even_positions[0])
print("Even scores:", scores[even_positions])

# np.where(condition, value_if_true, value_if_false) replaces values by condition
result = np.where(scores >= 80, "Pass", "Needs improvement")
print("Result:", result)
```

## Mathematical Operation

```python
arr1=np.array([15,20,25,30])
arr2=np.array([5,10,15,10])

print("Addition:",np.add(arr1,arr2),"\n")
print("Subtraction:",np.subtract(arr1,arr2),"\n")
print("Multiplication:",np.multiply(arr1,arr2),"\n")
print("Division:",np.divide(arr1,arr2),"\n")

print("Dot Product:",np.dot(arr1,arr2),"\n")
print("Power:",np.power(arr1,2),"\n")
print("Square Root:",np.sqrt(arr2),"\n")
```
